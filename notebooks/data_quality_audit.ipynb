```python
import pandas as pd
import glob
```

```python
dfs = []
for file in glob.glob('../data/raw/data_for_2024/*.csv'):
    df = pd.read_csv(file)
    dfs.append(df)

for file in glob.glob('../data/raw/data_for_2025/*.csv'):
    df = pd.read_csv(file,low_memory=False)
    dfs.append(df)

combined = pd.concat(dfs, ignore_index=True)
combined.shape
```

```python
new_combined = combined[combined['Origin'] == 'SEA']
print(new_combined.shape)
null_counts = new_combined.isna().sum()
cols_of_interest = [
    'ArrDelay',
    'FlightDate',
    'Year',
    'Quarter',
    'Month',
    'DayofMonth',
    'DayOfWeek',
    'IATA_CODE_Reporting_Airline',
    'Flight_Number_Reporting_Airline',
    'Origin',
    'Dest',
    'CRSDepTime',
    'CRSArrTime',
    'CRSElapsedTime',
    'Distance'
]
print(null_counts.loc[cols_of_interest])


```

```python
duplicate_criteria = [
    'FlightDate',
    'IATA_CODE_Reporting_Airline',
    'Flight_Number_Reporting_Airline',
    'Origin',
    'Dest'
]

print(new_combined.duplicated(subset=duplicate_criteria).sum())
```
