```python
#Loading all the csv's into one df
import pandas as pd
import glob

dfs = []
for file in glob.glob('../data/raw/data_for_2024/*.csv'):
    df = pd.read_csv(file,low_memory=False)
    dfs.append(df)

for file in glob.glob('../data/raw/data_for_2025/*.csv'):
    df = pd.read_csv(file,low_memory=False)
    dfs.append(df)

combined_unfiltered = pd.concat(dfs, ignore_index=True)
combined_unfiltered.shape
```

```python
#Filtering to only SEA-origin flights
combined_filtered = combined_unfiltered[combined_unfiltered['Origin'] == 'SEA']
combined_filtered.shape
```

```python
#Now to calculate cancellation and diversion rate amongs SEA-origin flights
cancelled_flights = combined_filtered['Cancelled'].sum()
diverted_flights = combined_filtered['Diverted'].sum()
cancellation_rate = (cancelled_flights/combined_filtered.shape[0]) * 100
diversion_rate = (diverted_flights/combined_filtered.shape[0]) * 100
print(cancellation_rate)
print(diversion_rate)

#So the cancellation rate is  approximately 0.96% while the diversion rate is approximately 0.27%,together the rate of excluded outcomes from the model is 1.23%
```

```python
#Dropping Null values based on target since the target had 4069 null values due to cancelled or diverted flights which is consistent with the scope of this project
combined_filtered = combined_filtered.dropna(subset=['ArrDelay'])
combined_filtered.shape
```

```python
#Checking if the cleaned dataset has no null values in all the features
null_counts = combined_filtered.isna().sum()
cols_of_interest = [
    'ArrDelay',
    'FlightDate',
    'Year',
    'Quarter',
    'Month',
    'DayofMonth',
    'DayOfWeek',
    'IATA_CODE_Reporting_Airline',
    'Flight_Number_Reporting_Airline',
    'Origin',
    'Dest',
    'CRSDepTime',
    'CRSArrTime',
    'CRSElapsedTime',
    'Distance'
]
print(null_counts.loc[cols_of_interest])
#Confirmed,no null counts based on the features now exists in the cleaned dataset
```

```python
#Saving cleaned df in csv format to `data/interim`

combined_filtered.to_csv('../data/interim/seattle_ontime_clean.csv')
```
